```python
class Banana():
    def __init__(self,name):
        self.name = name

    def info(self):
        return f"100 calories {self.name}"

class Apple():
    def __init__(self,name):
        self.name = name

    def info(self):
        return f"150 calories {self.name}"

```

```python
banana = Banana("banana")
```

```python
apple = Apple("apple")
```

```python
banana.info()
```

```python
fruitList = [apple, banana]
```

```python
for i in fruitList:
    print(i.info())
```

# Encapsulation

```python
class Phone():
    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def info(self):
        print(f"{self.name} {self.__price}")

    def changePrice(self,price):
        self.__price = price
```

```python
iphone = Phone("iphone",1400)
```

```python
iphone.info()
```

```python
iphone.__price = 400
```

```python
iphone.info()
```

```python
iphone.changePrice(300)
```

```python
iphone.info()
```

# Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class Car(ABC):

    @abstractmethod
    def maxSpeed(self):
        pass
```

```python
class Tesla(Car):
    def maxSpeed(self):
        print("200 km")
```

```python
tesla = Tesla()
```

```python
tesla.maxSpeed()
```
