```python
def divideNumber(num):
    return num / 2
```

```python
myList = [10,20,30,40] 
```

```python
myResultList = []
```

```python
for num in myList:
    myResultList.append(divideNumber(num))
```

```python
myResultList
```

```python
list(map(divideNumber,myList))
```

```python
class Musician():
    def __init__(self,name):
        self.name = name
        print("musician class")

    def  test1(self):
        print("test1")

    def  test2(self):
        print("test2")
```

```python
halit = Musician("halit")
```

```python
class MusicianPlus(Musician):
    def __init__(self,name):
        Musician.__init__(self,name)
        print("musicianplus class")
        
```

```python
bektas = MusicianPlus("bektas")
```
